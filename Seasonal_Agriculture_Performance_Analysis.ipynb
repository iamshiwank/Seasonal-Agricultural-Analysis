Objective: analyze how agricultural performance varies across Kharif, Rabi and Zaid seasons, and identify patterns in environmental conditions, resource usage and economic outcomes.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

file_path = 'seasonal_agriculture_performance_dataset.xlsx'
df = pd.read_excel(file_path)
df.head()
```

## 1. Dataset Overview

```python
print('Rows, Columns:', df.shape)
print('\nColumns:')
print(df.columns.tolist())
print('\nData types:')
print(df.dtypes)
print('\nDuplicate rows:', df.duplicated().sum())
```

## 2. Missing Values and Cleaning

```python
missing = df.isna().sum().sort_values(ascending=False)
print(missing[missing>0])

# Numeric missing values are imputed with the median to preserve the dataset size.
for col in df.select_dtypes(include=np.number).columns:
    if df[col].isna().any():
        df[col] = df[col].fillna(df[col].median())
print('Missing values after cleaning:', int(df.isna().sum().sum()))
```

## 3. Categorical Overview

```python
for col in ['Season','Crop','State','District','Irrigation_Method']:
    print(f'\n{col}: {df[col].nunique()} unique values')
    print(df[col].value_counts())
```

## 4. Descriptive Statistics

```python
df.describe().T.round(2)
```

## 5. Key Analytical Questions
1. How does agricultural performance vary across seasons?
2. How do production, revenue and profit differ by season?
3. How do environmental conditions and resource usage vary across seasons?
4. Which crops show the strongest yield and profitability?
5. Which irrigation methods are associated with better yield, water efficiency and profit?

## 6. Seasonal Performance

```python
season_metrics = df.groupby('Season').agg(
    Avg_Yield=('Yield_Tonnes_Ha','mean'),
    Avg_Production=('Production_Tonnes','mean'),
    Avg_Revenue=('Revenue_INR','mean'),
    Avg_Profit=('Profit_INR','mean'),
    Avg_Water_Used=('Water_Used_m3','mean'),
    Avg_Water_Efficiency=('Water_Efficiency_t_per_1000m3','mean'),
    Avg_Risk=('Disease_Pest_Risk_pct','mean')
).round(2)
season_metrics
```

```python
plt.figure(figsize=(8,5))
sns.barplot(data=df, x='Season', y='Yield_Tonnes_Ha', estimator='mean', errorbar=None)
plt.title('Average Yield by Season')
plt.ylabel('Yield (Tonnes/Ha)')
plt.xlabel('Season')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.barplot(data=df, x='Season', y='Profit_INR', estimator='mean', errorbar=None)
plt.title('Average Profit by Season')
plt.ylabel('Profit (INR)')
plt.xlabel('Season')
plt.tight_layout()
plt.show()
```

## 7. Environmental and Resource Comparison

```python
env_resource = df.groupby('Season')[['Rainfall_mm','Avg_Temperature_C','Humidity_pct','Soil_Moisture_pct','Fertilizer_kg_ha','Water_Used_m3','Water_Efficiency_t_per_1000m3']].mean().round(2)
env_resource
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(data=df, x='Season', y='Disease_Pest_Risk_pct')
plt.title('Disease/Pest Risk Distribution by Season')
plt.ylabel('Risk (%)')
plt.tight_layout()
plt.show()
```

## 8. Crop Performance

```python
crop_metrics = df.groupby('Crop').agg(Avg_Yield=('Yield_Tonnes_Ha','mean'), Avg_Profit=('Profit_INR','mean'), Avg_Production=('Production_Tonnes','mean')).sort_values('Avg_Yield', ascending=False).round(2)
crop_metrics
```

```python
plt.figure(figsize=(10,5))
plot_data = crop_metrics.reset_index().sort_values('Avg_Yield')
sns.barplot(data=plot_data, y='Crop', x='Avg_Yield', errorbar=None)
plt.title('Average Yield by Crop')
plt.xlabel('Yield (Tonnes/Ha)')
plt.ylabel('Crop')
plt.tight_layout()
plt.show()
```

## 9. Irrigation Analysis

```python
irrigation_metrics = df.groupby('Irrigation_Method')[['Yield_Tonnes_Ha','Water_Efficiency_t_per_1000m3','Profit_INR']].mean().sort_values('Yield_Tonnes_Ha', ascending=False).round(2)
irrigation_metrics
```

## 10. Relationships with Yield

```python
corr_cols = ['Yield_Tonnes_Ha','Profit_INR','Water_Used_m3','Rainfall_mm','Avg_Temperature_C','Soil_Moisture_pct','Nitrogen_kg_ha','Phosphorus_kg_ha','Potassium_kg_ha','Fertilizer_kg_ha','Pesticide_Litre_ha','Seed_Quality_Score','Disease_Pest_Risk_pct']
yield_corr = df[corr_cols].corr()['Yield_Tonnes_Ha'].sort_values(ascending=False)
yield_corr
```

```python
plt.figure(figsize=(8,6))
sns.heatmap(df[corr_cols].corr(), cmap='coolwarm', center=0)
plt.title('Correlation Matrix of Key Variables')
plt.tight_layout()
plt.show()
```

## 11. Seasonal Crop Comparison

```python
season_crop = df.pivot_table(index='Crop', columns='Season', values='Yield_Tonnes_Ha', aggfunc='mean').round(2)
season_crop
```

## 12. Key Findings
Based on the observed data, Kharif has the highest average yield, production, revenue and profit among the three seasons. Zaid has the lowest average profit and the lowest average water efficiency. Sugarcane has a much higher average yield than the other crops, so crop-level comparisons should be interpreted with awareness of crop differences. Drip irrigation shows the highest average yield and profit among irrigation methods, while Rainfed has the highest average water-efficiency value. Yield has a moderate positive association with profit and water used, while the listed environmental/input variables show relatively weak linear correlations with yield in this dataset. These are descriptive findings, not proof of causation.

## 13. Conclusion and Recommendations
- Seasonal conditions are associated with meaningful differences in agricultural outcomes.
- Kharif appears strongest on average across major performance measures in this dataset.
- Water efficiency should be considered alongside yield and profit when evaluating farming performance.
- Drip irrigation is associated with strong average yield and profit and merits further investigation.
- Crop type is an important consideration because yield and profitability vary substantially across crops.
- Future analysis could use predictive modelling, more detailed regional analysis, time-series data and weather/market data to improve planning.
